```python
# =============================================
# INSIDER THREAT DETECTION - ML-förberedelser
# DYNAMISK VERSION - Frågar efter allt i terminalen
# Skapar samma variabler som originalkoden (X, y, df)
# =============================================

print("="*70)
print("INSIDER THREAT DETECTION - ML-förberedelser")
print("="*70)

# =============================================
# Steg 1: Installera bibliotek
# =============================================
print("\nSTEG 1: Installerar bibliotek (om det behövs)...")
%pip install snowflake-connector-python pandas scikit-learn -q
print("   KLAR")

# =============================================
# Steg 2: Importera bibliotek
# =============================================
print("\nSTEG 2: Importerar bibliotek...")
import snowflake.connector as sf
import pandas as pd
from getpass import getpass
print("   KLAR")

# =============================================
# Steg 3: Fråga efter Snowflake-uppgifter
# =============================================
print("\n" + "="*70)
print("STEG 3: Ange Snowflake-anslutningsuppgifter")
print("="*70)

print("""
HUR DU HITTAR DINA SNOWFLAKE-UPPGIFTER:

ACCOUNT:   Logga in på Snowflake -> titta i URL: https://app.snowflake.com/DITT_ACCOUNT/...
           Exempel: 'RQKXOYX-PV94267'

USER:      Ditt användarnamn i Snowflake
           Exempel: 'ARSENIY'

PASSWORD:  Ditt Snowflake-lösenord (syns inte när du skriver)

WAREHOUSE: Välj ett warehouse som ska användas för beräkningar
           Exempel: 'COMPUTE_WH', 'DEV_WH'

DATABASE:  Databasen som innehåller din tabell
           Exempel: 'INSIDER_THREAT_DB'

SCHEMA:    Schemat som innehåller din tabell
           Exempel: 'DBT_ABUZYAN', 'PUBLIC'

OBS: För att få 14 features (kolumner) behöver du välja rätt schema och tabell.
     Tabellen 'FCT_FEATURES' innehåller 14 numeriska kolumner.
     Tabellen 'RAW_ACCESS_LOGS' innehåller bara 11 numeriska kolumner.
""")

print("\nAnge dina Snowflake-uppgifter:")

account = input("   ACCOUNT (t.ex. RQKXOYX-PV94267): ").strip()
user = input("   USER (ditt användarnamn): ").strip()
password = getpass("   PASSWORD (syns inte): ")
warehouse = input("   WAREHOUSE (t.ex. COMPUTE_WH): ").strip()
database = input("   DATABASE (t.ex. INSIDER_THREAT_DB): ").strip()
schema = input("   SCHEMA (t.ex. DBT_ABUZYAN): ").strip()

print("\n   Ansluter till Snowflake...")
conn = sf.connect(
    account=account,
    user=user,
    password=password,
    warehouse=warehouse,
    database=database,
    schema=schema
)
print("   Anslutning lyckades!")

# =============================================
# Steg 4: Fråga efter tabellnamn
# =============================================
print("\n" + "="*70)
print("STEG 4: Välj tabell att hämta data från")
print("="*70)

print("""
VIKTIGT OM TABELLER:

FCT_FEATURES   - Denna tabell innehåller 14 numeriska kolumner (features)
                 Detta är RÄTT tabell för ML-modellen.
                 Ger dig 14 features som modellen förväntar sig.

RAW_ACCESS_LOGS - Denna tabell innehåller ENDAST 11 numeriska kolumner
                 Detta är FEL tabell för ML-modellen.
                 Ger dig 11 features, vilket orsakar fel i A/B-testet.

REKOMMENDATION: Välj 'FCT_FEATURES' om den finns i listan nedan.
                Om den inte finns, kontrollera att du är i rätt schema.
""")

# Visa tillgängliga tabeller
cursor = conn.cursor()
cursor.execute("SHOW TABLES")
tables = cursor.fetchall()
print("\n   Tillgängliga tabeller i databasen:")
for t in tables[:10]:
    print(f"      - {t[1]}")
if len(tables) > 10:
    print(f"      ... och {len(tables)-10} fler")

print("\n")
print("Vilken tabell ska du välja?")
print("   - Om du ser 'FCT_FEATURES' -> skriv in det namnet (REKOMMENDERAS)")
print("   - Om du ser 'RAW_ACCESS_LOGS' -> detta ger bara 11 features (EJ REKOMMENDERAT)")
print("   - Annars skriv namnet på tabellen som innehåller dina features")

table_name = input("\nAnge tabellnamn: ").strip()

# Varna om användaren väljer fel tabell
if table_name.upper() == 'RAW_ACCESS_LOGS':
    print("\n   *** VARNING: Du valde RAW_ACCESS_LOGS! ***")
    print("   Denna tabell har bara 11 features (kolumner).")
    print("   Dina modeller är tränade med 14 features.")
    print("   A/B-testet kommer att misslyckas om du använder denna tabell.")
    print("")
    confirm = input("   Vill du fortsätta ändå? (ja/nej): ").strip().lower()
    if confirm != 'ja':
        print("   Välj om och skriv 'FCT_FEATURES' istället.")
        raise SystemExit("Anvandaren valde att avbryta. Välj ratt tabell nasta gang.")

# =============================================
# Steg 5: Hämta data
# =============================================
print("\n" + "="*70)
print("STEG 5: Hämtar data från Snowflake")
print("="*70)

# Kontrollera antal rader
cursor.execute(f"SELECT COUNT(*) FROM {table_name}")
row_count = cursor.fetchone()[0]
print(f"   Hittade {row_count:,} rader i {table_name}")

# Fråga om limit
print("\n   Vill du begränsa antalet rader? (ja/nej)")
print("   Rekommendation: Svara 'nej' för att hämta all data.")
print("   (Vid 'nej' hämtas alla rader)")
limit_choice = input("   -> ").strip().lower()

if limit_choice == 'ja':
    limit_count = input("   Antal rader att hämta (t.ex. 1000): ").strip()
    limit_count = int(limit_count) if limit_count else 1000
    query = f"SELECT * FROM {table_name} LIMIT {limit_count}"
    print(f"   Hämtar {limit_count} rader...")
else:
    query = f"SELECT * FROM {table_name}"
    print("   Hämtar alla rader...")

# Hämta data
df = pd.read_sql(query, conn)
print(f"Hämtade {len(df):,} rader och {len(df.columns)} kolumner")

# Stäng anslutningen
conn.close()
print("Anslutning stängd.")

# =============================================
# Steg 6: Förbered data för ML
# =============================================
print("\n" + "="*70)
print("STEG 6: Förbereder data för ML")
print("="*70)

# Kontrollera att ATTACK_TYPE finns
if 'ATTACK_TYPE' not in df.columns:
    print(f"   VARNING: Kolumnen 'ATTACK_TYPE' hittades inte!")
    print(f"   Tillgängliga kolumner: {list(df.columns)}")
    
    print("\n   Ange vilken kolumn som ska användas som målvariabel:")
    for i, col in enumerate(df.columns):
        print(f"      {i+1}. {col}")
    target_col = input("   -> ").strip()
    
    if target_col.isdigit() and 1 <= int(target_col) <= len(df.columns):
        target_col = df.columns[int(target_col)-1]
else:
    target_col = 'ATTACK_TYPE'
    print(f"   Hittade målvariabel: {target_col}")

# Välj target (y)
y = df[target_col]

# Välj bara numeriska kolumner för ML
numerical_cols = df.select_dtypes(include=['number']).columns
X = df[numerical_cols]

print(f"\n   Features (X): {len(X.columns)} numeriska kolumner")
print(f"   Target (y): {y.nunique()} unika varden")

# Kontrollera antalet features
if len(X.columns) == 14:
    print("\n   *** PERFEKT! X har 14 features. ***")
    print("   Detta är RÄTT antal för dina modeller.")
elif len(X.columns) == 11:
    print("\n   *** VARNING: X har bara 11 features! ***")
    print("   Dina modeller är tränade med 14 features.")
    print("   Du har valt fel tabell (RAW_ACCESS_LOGS istället för FCT_FEATURES).")
    print("   A/B-testet kommer att misslyckas.")
    print("\n   För att fixa: Kör om denna cell och välj 'FCT_FEATURES'.")
else:
    print(f"\n   OBS: X har {len(X.columns)} features.")

# Visa de första raderna
print("\nForsta 5 raderna av datan:")
print(df.head())

# =============================================
# Steg 7: Dataöversikt
# =============================================
print("\n" + "="*70)
print("STEG 7: Dataöversikt")
print("="*70)

print(f"\n   Dataframe dimensioner: {df.shape}")
print(f"   Features (numeriska): {list(X.columns)}")
print(f"   Target distribution:")
print(y.value_counts().head(10))

# Kontrollera om det är binär klassificering (attack vs normal)
if y.dtype == 'object':
    if 'normal' in y.values:
        attack_pct = (y != 'normal').mean() * 100
        normal_pct = (y == 'normal').mean() * 100
        print(f"\n   Klassfordelning:")
        print(f"      Normal: {normal_pct:.1f}%")
        print(f"      Attack: {attack_pct:.1f}%")

# =============================================
# Steg 8: Sammanfattning och nästa steg
# =============================================
print("\n" + "="*70)
print("SAMMANFATTNING")
print("="*70)

print(f"""
Vald tabell: {table_name}
Antal features (X): {len(X.columns)} (borde vara 14)
Antal rader: {len(df):,}
Attack-rate: {attack_pct:.1f}% (om 'normal' finns)

VARIABLER SOM FINNS I WORKSPACE:
- df  -> Hela datan (DataFrame)
- X   -> Features (numeriska kolumner) - KLAR FOR ML
- y   -> Target (malvariabel) - KLAR FOR ML
""")

if len(X.columns) == 14:
    print("""
*** STATUS: KLAR FOR ML-MODELLEN ***
Du har 14 features - RATT antal!

NASTA STEG:
1. Fortsätt till Steg 9 (Importera MLflow)
2. Kör Steg 10 (Bygg ML-modell)
3. Kör Steg 11 (Logga till MLflow)
4. Kör A/B-test för att jamfora modeller
""")
elif len(X.columns) == 11:
    print("""
*** STATUS: PROBLEM MED FEATURES ***
Du har bara 11 features - FEL antal!

PROBLEM: Du valde fel tabell (RAW_ACCESS_LOGS)
LOSNING: Kör om denna cell och valj 'FCT_FEATURES'

NASTA STEG:
1. Kör om denna cell
2. Välj 'FCT_FEATURES' som tabellnamn
3. Fortsätt sedan som vanligt
""")
else:
    print("""
*** STATUS: OKANDE ANTAL FEATURES ***
Kontrollera att du har valt ratt tabell.

NASTA STEG:
1. Verifiera att tabellen innehaller alla 14 numeriska kolumner
2. Fortsätt med forsiktighet
""")

print("\n" + "="*70)
print("KLART! Fortsätt till nästa cell.")
print("="*70)
```

# **step_10_random_forest**

- **Vad gor koden?**
- Trader och jamfor tva olika maskininlarningsmodeller for att upptacka insiderhot. Valjer den modell som fungerar bast.
- **Varfor behovs den?**
- Vi vet inte i forvag vilken modell som fungerar bast - måste testa
- Isolation Forest upptacker avvikelser (bra for ovanliga attacker)
- Random Forest lär sig från monster (bra for vanliga attacktyper)
- Genom att jamfora får vi den basta modellen for just var data
- **Hur fungerar den?**
- Skapar binär etikett - Gör om 'attack' till 1 och 'normal' till 0
- Delar upp data - 80% for träning, 20% for test
- Skalar features - Gör alla numeriska varden jamforbara
- Trader Isolation Forest - Testar olika inställningar, valjer den basta
- Trader Random Forest - Testar olika inställningar, valjer den basta
- Jamfor modellerna - Ser vilken som har hogst F1-score
- Valjer vinnare - Sparar den basta modellen
- Visar feature importance - Visar vilka matvarden som viktigast (for Random Forest)
- Sparar i MLflow - Loggar alla resultat och modellen
- **Viktiga begrepp:**
- Precision = Andel av flaggade som verkligen var attacker (99.1% i ditt fall)
- Recall = Andel attacker som upptäcktes (96.7% i ditt fall)
- F1-score = Balans mellan precision och recall (0.979 i ditt fall)
- Confusion Matrix = Visar hur manga ratt/fel modellen gjorde
- **Resultat efter körning:**
- Random Forest blev bästa modell (F1-score: 0.979)
- Precision: 99.1% - nar modellen flaggar attack har den ratt i 99 av 100 fall
- Recall: 96.7% - modellen upptacker 97 av 100 attacker
- Modellen och scaler ar sparade i MLflow

```python
# =============================================
# STEG 10: FÖRBÄTTRAD ML-MODELL (BÄTTRE RESULTAT)
# =============================================

print("🚀 Steg 10 (Förbättrad): Bygger optimerad ML-modell...")

# Importera bibliotek
from sklearn.ensemble import IsolationForest, RandomForestClassifier
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, precision_score, recall_score, f1_score, roc_auc_score
import numpy as np
import pandas as pd

# =============================================
# Kontrollera att X och y finns
# =============================================
try:
    print(f"📊 Data från Steg 8: {X.shape[0]} rader, {X.shape[1]} features")
    print(f"🎯 Målvariabel: {y.nunique()} unika attacktyper")
except NameError:
    print("❌ X och y saknas! Kör först Steg 8-koden.")
    raise SystemExit("Kör Steg 8 innan Steg 10!")

# =============================================
# Steg 1: Skapa binär etikett (attack = 1, normal = 0)
# =============================================
y_binary = (y != 'normal').astype(int)
print(f"\n📊 Klassfördelning:")
print(f"   Normal (0): {(y_binary == 0).sum()} rader ({(y_binary == 0).mean()*100:.1f}%)")
print(f"   Attack (1): {(y_binary == 1).sum()} rader ({(y_binary == 1).mean()*100:.1f}%)")

# =============================================
# Steg 2: Dela upp data
# =============================================
X_train, X_test, y_train, y_test = train_test_split(
    X, y_binary, test_size=0.2, random_state=42, stratify=y_binary
)
print(f"\n📊 Träningsdata: {len(X_train)} rader")
print(f"   Testdata: {len(X_test)} rader")

# =============================================
# Steg 3: Skala features
# =============================================
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# =============================================
# Steg 4: OPTIMERAD ISOLATION FOREST (auto contamination)
# =============================================
print("\n" + "="*50)
print("🔍 MODELL 1: Optimerad Isolation Forest")
print("="*50)

# Beräkna optimal contamination (andel attacker i datan)
optimal_contamination = (y_binary == 1).mean() * 1.5  # 1.5x för att fånga fler
optimal_contamination = min(optimal_contamination, 0.5)  # Max 50%

print(f"   Optimal contamination: {optimal_contamination:.3f}")

# Testa flera parametrar
iso_params = [
    {'contamination': [0.1, 0.15, 0.2, 0.25, 0.3, optimal_contamination],
     'n_estimators': [50, 100, 200],
     'max_samples': ['auto', 0.7, 0.8]}
]

best_iso = None
best_iso_f1 = 0

for cont in [0.1, 0.15, 0.2, 0.25, 0.3, optimal_contamination]:
    for n_est in [50, 100, 200]:
        iso = IsolationForest(
            contamination=cont,
            n_estimators=n_est,
            random_state=42,
            max_samples='auto'
        )
        iso.fit(X_train_scaled)
        pred = iso.predict(X_test_scaled)
        pred_binary = (pred == -1).astype(int)
        f1 = f1_score(y_test, pred_binary)
        
        if f1 > best_iso_f1:
            best_iso_f1 = f1
            best_iso = iso
            best_iso_params = (cont, n_est)

print(f"   ✅ Bästa Isolation Forest:")
print(f"      contamination={best_iso_params[0]}, n_estimators={best_iso_params[1]}")
print(f"      F1-score: {best_iso_f1:.3f}")

# Gör förutsägelser med bästa modellen
iso_pred = best_iso.predict(X_test_scaled)
iso_pred_binary = (iso_pred == -1).astype(int)

# =============================================
# Steg 5: RANDOM FOREST (övervakad modell)
# =============================================
print("\n" + "="*50)
print("🌲 MODELL 2: Random Forest (övervakad)")
print("="*50)

# Hyperparameter tuning för Random Forest
rf_params = {
    'n_estimators': [100, 200],
    'max_depth': [10, 20, None],
    'class_weight': ['balanced', None]
}

rf = RandomForestClassifier(random_state=42)
grid_search = GridSearchCV(rf, rf_params, cv=3, scoring='f1', n_jobs=-1)
grid_search.fit(X_train_scaled, y_train)

best_rf = grid_search.best_estimator_
print(f"   ✅ Bästa Random Forest:")
print(f"      {grid_search.best_params_}")
print(f"      Bästa CV F1-score: {grid_search.best_score_:.3f}")

# Gör förutsägelser
rf_pred = best_rf.predict(X_test_scaled)

# =============================================
# Steg 6: JÄMFÖR MODELLERNA
# =============================================
print("\n" + "="*50)
print("📊 JÄMFÖRELSE MELLAN MODELLER")
print("="*50)

# Beräkna metrics för båda modellerna
models = {
    'Isolation Forest': iso_pred_binary,
    'Random Forest': rf_pred
}

results_df = pd.DataFrame()

for model_name, pred in models.items():
    precision = precision_score(y_test, pred)
    recall = recall_score(y_test, pred)
    f1 = f1_score(y_test, pred)
    
    cm = confusion_matrix(y_test, pred)
    
    results_df.loc[model_name, 'Precision'] = f"{precision:.3f}"
    results_df.loc[model_name, 'Recall'] = f"{recall:.3f}"
    results_df.loc[model_name, 'F1-score'] = f"{f1:.3f}"
    results_df.loc[model_name, 'True Positives'] = cm[1,1]
    results_df.loc[model_name, 'False Positives'] = cm[0,1]
    results_df.loc[model_name, 'False Negatives'] = cm[1,0]

print("\n📈 Resultatjämförelse:")
print(results_df.to_string())

# =============================================
# Steg 7: Välj bästa modell
# =============================================
if f1_score(y_test, rf_pred) > best_iso_f1:
    best_model = best_rf
    best_model_name = "Random Forest"
    best_pred = rf_pred
    print(f"\n✅ Bästa modell: Random Forest (F1: {f1_score(y_test, rf_pred):.3f})")
else:
    best_model = best_iso
    best_model_name = "Isolation Forest"
    best_pred = iso_pred_binary
    print(f"\n✅ Bästa modell: Isolation Forest (F1: {best_iso_f1:.3f})")

# =============================================
# Steg 8: Detaljerad analys av bästa modellen
# =============================================
print("\n" + "="*50)
print(f"🔍 DETALJERAD ANALYS: {best_model_name}")
print("="*50)

cm = confusion_matrix(y_test, best_pred)
tn, fp, fn, tp = cm.ravel()

print(f"\n   Confusion Matrix:")
print(f"   ┌─────────────────┐")
print(f"   │  Normal   Attack│")
print(f"   ├─────────────────┤")
print(f"   │  {tn:3d}       {fp:3d}   │ ← Normal (sant/falskt)")
print(f"   │  {fn:3d}       {tp:3d}   │ ← Attack (falskt/sant)")
print(f"   └─────────────────┘")

precision = precision_score(y_test, best_pred)
recall = recall_score(y_test, best_pred)
f1 = f1_score(y_test, best_pred)

print(f"\n   📊 Metrics:")
print(f"   ✅ Precision: {precision:.3f} ({precision*100:.1f}% av flaggade var attacker)")
print(f"   ✅ Recall:    {recall:.3f} ({recall*100:.1f}% av attacker flaggades)")
print(f"   ✅ F1-score:  {f1:.3f}")

# =============================================
# Steg 9: Feature importance (för Random Forest)
# =============================================
if best_model_name == "Random Forest":
    print("\n   📊 Feature Importance (viktigaste features för modellen):")
    feature_importance = pd.DataFrame({
        'feature': X.columns,
        'importance': best_model.feature_importances_
    }).sort_values('importance', ascending=False)
    
    for i, row in feature_importance.head(5).iterrows():
        bar = "█" * int(row['importance'] * 50)
        print(f"      {row['feature']:25s}: {row['importance']:.3f} {bar}")

# =============================================
# Steg 10: LOGGA TILL MLflow
# =============================================
print("\n📊 Loggar experiment till MLflow...")

with mlflow.start_run(run_name="Improved_ML_Models"):
    # Logga båda modellernas resultat
    mlflow.log_param("best_model", best_model_name)
    
    for model_name, pred in models.items():
        mlflow.log_metric(f"{model_name}_precision", precision_score(y_test, pred))
        mlflow.log_metric(f"{model_name}_recall", recall_score(y_test, pred))
        mlflow.log_metric(f"{model_name}_f1", f1_score(y_test, pred))
    
    # Logga bästa modellen
    if best_model_name == "Random Forest":
        mlflow.sklearn.log_model(best_model, "best_random_forest_model")
    else:
        mlflow.sklearn.log_model(best_model, "best_isolation_forest_model")
    
    # Logga scaler
    import joblib
    joblib.dump(scaler, "scaler.pkl")
    mlflow.log_artifact("scaler.pkl")
    
    print("   ✅ Experiment loggat till MLflow!")

# =============================================
# SAMMANFATTNING
# =============================================
print("\n" + "="*55)
print("✅ STEG 10 (FÖRBÄTTRAD) ÄR KLART!")
print("="*55)
print(f"\n🏆 Bästa modell: {best_model_name}")
print(f"\n📈 Slutgiltiga resultat:")
print(f"   🎯 Precision: {precision:.3f} ({precision*100:.1f}%)")
print(f"   🎯 Recall:    {recall:.3f} ({recall*100:.1f}%)")
print(f"   🎯 F1-score:  {f1:.3f}")

if f1 > 0.6:
    print("\n🎉 Utmärkt resultat! Modellen är redo för produktion!")
elif f1 > 0.5:
    print("\n👍 Bra resultat! Modellen kan användas med viss övervakning.")
else:
    print("\n⚠️ Resultatet kan förbättras. Överväg mer feature engineering.")

print("\n💡 Nästa steg: Steg 11 - Skapa Databricks Job för automatisering")
```

## MULTI-AGENT SYSTEM

- **Vad gor koden?**
- Anvander fyra AI-agenter (baserade pa Groq API) som samarbetar for att analysera och forklara modellens resultat. Varje agent har en specifik uppgift.
- **Varfor behovs den?**
- En ensam AI kan missa viktiga detaljer
- Olika experter ser olika problem (ML-tekniker vs sakerhetsexpert)
- Flera agenter kan granska och forbattra varandras arbete
- Resultatet blir mer trovandigt och anvandbart for ledningen
- **Hur fungerar den?**
- Väljer API - Du kan anvanda Groq (riktig AI) eller simulering (demo)
- Hamtar modellresultat - Precision, recall, F1-score fran Steg 10
- Agent 1 (ML-analytiker) - Analyserar modellens styrkor och svagheter
- Agent 2 (Sakerhetsexpert) - Tolkar hotbilden utifran ML-analysen
- Agent 3 (Rapportskrivare) - Skriver en rapport till ledningen
- Agent 4 (Kvalitetsgranskare) - Forbattrar rapporten och gor den mer koncis
- Slutrapport - Visar den fardiga analysen
- **De fyra agenternas roller:**
- Agent 1: ML-analytiker - Analyserar precision, recall och F1-score
- Agent 2: Sakerhetsexpert - Bedomer risker och ger rekommendationer
- Agent 3: Rapportskrivare - Sammanstaller till ledningsrapport
- Agent 4: Kvalitetsgranskare - Forbattrar och forkortar rapporten
- **Resultat efter körning:**
- En professionell analys av modellens prestanda
- Identifierade styrkor (hog precision) och svagheter (23 missade attacker)
- En konkret rekommendation for nasta steg
- Rapport klar att delas med ledningen

```python
# =============================================
# MULTI-AGENT SYSTEM: 4 agenter samarbetar
# =============================================
# ENDAST GROQ API (fungerar) + Simulering
# Varje agent har EN specifik uppgift
# =============================================

print("="*60)
print("🤖 MULTI-AGENT SYSTEM: Förklarar Random Forest-modellen")
print("="*60)

# =============================================
# Steg 1: Välj API
# =============================================
print("\n📋 Välj API för agenterna:")
print("   1. Groq (gratis, snabbast) ⭐")
print("   2. Simulering (ingen API, bara demo)")

api_choice = input("\n🔢 Ange siffra (1-2): ").strip()

from getpass import getpass
client = None
model_name = None

if api_choice == "1":
    print("\n🔑 Ange din Groq API-nyckel (syns inte):")
    api_key = getpass()
    
    if api_key:
        print("\n🔧 Installerar Groq...")
        %pip install groq -q
        
        from groq import Groq
        client = Groq(api_key=api_key)
        model_name = "llama-3.3-70b-versatile"
        print("✅ Groq redo!")
    else:
        print("❌ Ingen API-nyckel. Använder simulering.")
        api_choice = "2"

# =============================================
# Steg 2: Hämta modellresultat (från Steg 10)
# =============================================
try:
    from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix
    precision = precision_score(y_test, best_pred)
    recall = recall_score(y_test, best_pred)
    f1 = f1_score(y_test, best_pred)
    cm = confusion_matrix(y_test, best_pred)
    tn, fp, fn, tp = cm.ravel()
    
    print(f"\n📊 Modellresultat från Steg 10:")
    print(f"   Precision: {precision:.3f} ({precision*100:.1f}%)")
    print(f"   Recall: {recall:.3f} ({recall*100:.1f}%)")
    print(f"   F1-score: {f1:.3f}")
    print(f"   Falska positiva: {fp}")
    print(f"   Missade attacker: {fn}")
    
except NameError:
    print("❌ Kör Steg 10 först!")
    raise SystemExit()

# =============================================
# AGENT 1: ML-ANALYTIKER
# =============================================
print("\n" + "="*50)
print("👨‍🔬 AGENT 1: ML-analytiker – Analyserar modellens resultat")
print("="*50)

def agent_ml_analyst():
    prompt = f"""
    Du är en expert på maskininlärning. Analysera dessa resultat för en Random Forest-modell:
    
    - Precision: {precision*100:.1f}% (av flaggade, {precision*100:.1f}% var attacker)
    - Recall: {recall*100:.1f}% (av alla attacker, {recall*100:.1f}% upptäcktes)
    - F1-score: {f1:.3f}
    - Falska positiva: {fp}
    - Missade attacker: {fn}
    
    Svara i exakt detta format (max 3 meningar):
    STYRKA: [vad är bra?]
    SVAGHET: [vad kan förbättras?]
    SLUTSATS: [övergripande bedömning]
    """
    
    if api_choice == "1" and client:
        response = client.chat.completions.create(
            model=model_name,
            messages=[{"role": "user", "content": prompt}],
            temperature=0.3
        )
        return response.choices[0].message.content
    else:
        return """STYRKA: Hög precision (99.1%) innebär mycket få falska larm.
SVAGHET: Recall (96.7%) är bra men 23 attacker missades.
SLUTSATS: Modellen är utmärkt men kan finjusteras för att fånga fler attacker."""

ml_analysis = agent_ml_analyst()
print(f"\n📋 Agent 1 svarar:\n{ml_analysis}")

# =============================================
# AGENT 2: SÄKERHETSEXPERT
# =============================================
print("\n" + "="*50)
print("🔐 AGENT 2: Säkerhetsexpert – Tolkar hotbilden")
print("="*50)

def agent_security_expert(ml_analysis):
    prompt = f"""
    Du är en cybersäkerhetsexpert. Baserat på denna ML-analys:
    
    {ml_analysis}
    
    Svara i exakt detta format (max 3 meningar):
    RISK: [vad är den största risken?]
    UPPTÄCKT: [vilka attacker upptäcks bäst?]
    REKOMMENDATION: [en åtgärd direkt]
    """
    
    if api_choice == "1" and client:
        response = client.chat.completions.create(
            model=model_name,
            messages=[{"role": "user", "content": prompt}],
            temperature=0.3
        )
        return response.choices[0].message.content
    else:
        return """RISK: 23 missade attacker kan vara insiderhot som inte upptäcks.
UPPTÄCKT: Modellen är bäst på att upptäcka credential_sharing och unauthorized_access.
REKOMMENDATION: Samla in fler normala exempel för att minska falska positiva."""

security_analysis = agent_security_expert(ml_analysis)
print(f"\n📋 Agent 2 svarar:\n{security_analysis}")

# =============================================
# AGENT 3: RAPPORTSKRIVARE
# =============================================
print("\n" + "="*50)
print("📝 AGENT 3: Rapportskrivare – Sammanställer till ledning")
print("="*50)

def agent_report_writer(ml_analysis, security_analysis):
    prompt = f"""
    Du är en teknisk skribent. Skriv en kort rapport (max 100 ord) till ledningen.
    
    ML-analys: {ml_analysis}
    Säkerhetsanalys: {security_analysis}
    
    Använd professionell svenska. Inkludera en tydlig rekommendation.
    """
    
    if api_choice == "1" and client:
        response = client.chat.completions.create(
            model=model_name,
            messages=[{"role": "user", "content": prompt}],
            temperature=0.4
        )
        return response.choices[0].message.content
    else:
        return """RAPPORT: Random Forest-modellen uppnår 99.1% precision och 96.7% recall, vilket är utmärkt för insiderhot-detektering. Endast 6 falska larm och 23 missade attacker på 800 testfall. Rekommendation: Distribuera modellen i produktion med kontinuerlig övervakning."""

draft_report = agent_report_writer(ml_analysis, security_analysis)
print(f"\n📋 Agent 3 svarar:\n{draft_report}")

# =============================================
# AGENT 4: KVALITETSGRANSKARE
# =============================================
print("\n" + "="*50)
print("🔍 AGENT 4: Kvalitetsgranskare – Förbättrar rapporten")
print("="*50)

def agent_quality_reviewer(draft_report):
    prompt = f"""
    Du är en kvalitetsgranskare. Förbättra denna rapport:
    
    {draft_report}
    
    Krav:
    - Max 80 ord (mer koncist)
    - Lägg till en konkret nästa steg
    - Behåll all viktig fakta
    
    Svara ENDAST med den förbättrade rapporten.
    """
    
    if api_choice == "1" and client:
        response = client.chat.completions.create(
            model=model_name,
            messages=[{"role": "user", "content": prompt}],
            temperature=0.2
        )
        return response.choices[0].message.content
    else:
        return """Random Forest-modellen har 99.1% precision och 96.7% recall – utmärkt för insiderhot. Endast 6 falska larm på 800 testfall. Nästa steg: Distribuera med A/B-test i 2 veckor."""

final_report = agent_quality_reviewer(draft_report)

# =============================================
# SLUTSAMLING
# =============================================
print("\n" + "="*60)
print("✅ MULTI-AGENT SYSTEM – SLUTRAPPORT")
print("="*60)
print(f"\n📋 {final_report}")
print("\n" + "="*60)
print("🔍 Varför multi-agent är bättre:")
print("   1. Specialisering – varje agent har EN uppgift")
print("   2. Kedjeprocess – agent 2 förbättrar agent 1")
print("   3. Kvalitetskontroll – agent 4 granskar agent 3")
print("   4. Industri-standard – används av PGA TOUR, Workday, Accenture")
print("="*60)
```

## A/B-TESTNING

- **Vad gor koden?**
- Jamfor tva olika Random Forest-modeller som du redan har traderat och sparat i MLflow. Koden testar bada modellerna pa samma data och berattar vilken som presterar bast.
- **Varfor behovs den?**
- Nar du traderar en ny modell vet du inte om den ar battre an den gamla
- A/B-test jamfor modellerna pa ett rattvist satt (samma testdata)
- Simuleringen visar hur modellerna skulle fungera i produktion med riktig trafik
- Hjalper dig fatta datadrivna beslut om vilken modell som ska anvandas
- **Hur fungerar den - steg for steg:**
- _Steg 1: Kontrollerar att data finns_

```python
# =============================================
# STEG 12: FULL AUTOMATISERING MED SLACK-RAPPORT + LLM-ANALYS
# Dynamisk version - Använder miljövariabler för känslig information
# =============================================

print("="*70)
print("FULL AUTOMATISERING MED SLACK-RAPPORT + LLM-ANALYS")
print("="*70)

print("""
FORSTA DETTA:
- Denna kod hamtar en modell från MLflow, testar den, och skickar rapport till Slack
- AI-genererad analys från Groq Llama 3.3
- Alla resultat sparas i MLflow och CSV

VIKTIGT:
- Du behover ha kopt den dynamiska dataladdningscellen (Steg 1-8) forst
- X och y ska finnas i workspace
- Slack Webhook och Groq API är valfria (tryck Enter for att hoppa over)
""")

input("Tryck Enter for att fortsatta...")

# =============================================
# Steg 0: Fråga efter API-nycklar och URL:er (dolt)
# =============================================
print("\n" + "="*70)
print("KONFIGURATION")
print("="*70)

from getpass import getpass
import requests
import os

# Slack Webhook URL (dolt)
print("\nSLACK WEBHOOK URL (valfritt)")
print("   Sa har far du den: https://api.slack.com/apps -> din app -> Incoming Webhooks")
print("   (Tryck Enter for att hoppa over Slack-notiser)")
slack_webhook_url = getpass("   Klistra in din Slack Webhook URL (syns inte): ").strip()

# Groq API-nyckel (dolt)
print("\nGROQ API NYCKEL (valfritt for AI-analys)")
print("   Sa har far du den: console.groq.com -> API Keys -> Create API Key")
groq_api_key = getpass("   Ange din Groq API-nyckel (eller Enter for att hoppa over): ")

print("\nStartar automation...")

# =============================================
# Steg 1: Importera bibliotek
# =============================================
print("\nImporterar bibliotek...")
import mlflow
from mlflow.tracking import MlflowClient
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix
import joblib
from datetime import datetime

client = MlflowClient()

# =============================================
# Steg 2: Kontrollera data
# =============================================
print("\n" + "="*70)
print("STEG 2: Kontrollera data")
print("="*70)

if 'X' not in globals() or 'y' not in globals():
    print("   FEL: X och y saknas!")
    print("   Kör först den dynamiska dataladdningscellen (Steg 1-8).")
    raise SystemExit("Data saknas")

X_data = globals()['X']
y_data_raw = globals()['y']

if hasattr(y_data_raw, 'dtype') and y_data_raw.dtype == 'object':
    y_data = (y_data_raw != 'normal').astype(int)
    print("   Konverterade target till binär (1=attack, 0=normal)")
else:
    y_data = y_data_raw

print(f"   Data: {X_data.shape[0]} rader, {X_data.shape[1]} features")
print(f"   Attack-rate: {y_data.mean()*100:.1f}%")

# =============================================
# Steg 3: Dela upp data
# =============================================
print("\n" + "="*70)
print("STEG 3: Dela upp data")
print("="*70)

X_train, X_test, y_train, y_test = train_test_split(
    X_data, y_data, test_size=0.2, random_state=42, stratify=y_data
)
print(f"   Training: {len(X_train)} rader (80%)")
print(f"   Test: {len(X_test)} rader (20%)")

# =============================================
# Steg 4: Skala data
# =============================================
print("\n" + "="*70)
print("STEG 4: Skala features")
print("="*70)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
print("   Data skalad")

# =============================================
# Steg 5: Välj och ladda modell från MLflow
# =============================================
print("\n" + "="*70)
print("STEG 5: Hämta modell från MLflow")
print("="*70)

print("\nAnge din e-postadress (samma som i MLflow):")
user_email = input("   -> ").strip()
experiment_name = f"/Users/{user_email}/insider_threat_project"

experiment = client.get_experiment_by_name(experiment_name)

if experiment:
    runs = mlflow.search_runs(
        experiment_ids=[experiment.experiment_id],
        order_by=["start_time DESC"],
        max_results=5
    )
    
    if len(runs) > 0:
        print("\n   Senaste modellerna (med 14 features):")
        valid_models = []
        for i, row in runs.iterrows():
            try:
                # Snabbtest av features utan att ladda hela modellen
                model_test = mlflow.sklearn.load_model(f"runs:/{row['run_id']}/random_forest_model")
                n_features = model_test.n_features_in_
                if n_features == 14:
                    f1_val = row.get('metrics.f1_score', 'N/A')
                    if isinstance(f1_val, str) or pd.isna(f1_val):
                        f1_display = "N/A"
                    else:
                        f1_display = f"{float(f1_val):.3f}"
                    print(f"      {len(valid_models)+1}. {row.get('tags.mlflow.runName', 'Unnamed')[:35]} - F1: {f1_display}")
                    valid_models.append(row)
            except:
                continue
        
        if len(valid_models) > 0:
            print(f"\n   Välj modell (1-{len(valid_models)}):")
            choice = input("   -> ").strip()
            
            if choice.isdigit() and 1 <= int(choice) <= len(valid_models):
                selected = valid_models[int(choice)-1]
                run_id = selected['run_id']
                model = mlflow.sklearn.load_model(f"runs:/{run_id}/random_forest_model")
                print(f"   Laddade modell från Run ID: {run_id[:8]}...")
                
                # Beräkna prestanda
                y_pred = model.predict(X_test_scaled)
                precision = precision_score(y_test, y_pred)
                recall = recall_score(y_test, y_pred)
                f1 = f1_score(y_test, y_pred)
                cm = confusion_matrix(y_test, y_pred)
                tn, fp, fn, tp = cm.ravel()
                n_estimators = model.n_estimators
                max_depth = model.max_depth if model.max_depth else 10
            else:
                raise SystemExit("Ogiltigt val")
        else:
            print("   Inga modeller med 14 features hittades!")
            raise SystemExit("Träna först en modell med 14 features")
    else:
        raise SystemExit("Inga modeller hittades")
else:
    raise SystemExit(f"Hittade inte experimentet: {experiment_name}")

print(f"\n   MODELLPRESTANDA:")
print(f"      Precision: {precision:.3f} ({precision*100:.1f}%)")
print(f"      Recall: {recall:.3f} ({recall*100:.1f}%)")
print(f"      F1-score: {f1:.3f}")
print(f"      Confusion Matrix: TP={tp}, FP={fp}, FN={fn}, TN={tn}")

# =============================================
# Steg 6: Hämta historisk prestanda
# =============================================
print("\n" + "="*70)
print("STEG 6: Hämta historisk prestanda")
print("="*70)

historical_f1_list = []
latest_f1 = None

try:
    mlflow.set_experiment(experiment_name)
    experiment = client.get_experiment_by_name(experiment_name)
    
    if experiment:
        runs = mlflow.search_runs(
            experiment_ids=[experiment.experiment_id],
            order_by=["start_time DESC"],
            max_results=20
        )
        
        for _, row in runs.iterrows():
            if 'metrics.f1_score' in row and pd.notna(row['metrics.f1_score']):
                historical_f1_list.append(row['metrics.f1_score'])
        
        if historical_f1_list:
            latest_f1 = historical_f1_list[0]
            historical_avg_f1 = np.mean(historical_f1_list)
            historical_max_f1 = np.max(historical_f1_list)
            print(f"   Hittade {len(historical_f1_list)} historiska körningar")
            print(f"      Senaste F1: {latest_f1:.3f}")
            print(f"      Historisk snitt: {historical_avg_f1:.3f}")
except Exception as e:
    print(f"   Kunde inte hämta historik: {e}")

# =============================================
# Steg 7: Beräkna förändring
# =============================================
if latest_f1 is not None:
    change_percent = ((f1 - latest_f1) / latest_f1) * 100
    print(f"\n   Jämförelse med senaste modellen:")
    print(f"      Senaste F1: {latest_f1:.3f} -> Ny F1: {f1:.3f} -> Förändring: {change_percent:+.2f}%")
    
    if change_percent > 1:
        status = "FORBATTRING"
    elif change_percent < -1:
        status = "FORSAMRING"
    else:
        status = "STABIL"
else:
    change_percent = 0
    status = "FORSTA KORNINGEN"

# =============================================
# Steg 8: AI-analys (om Groq finns)
# =============================================
print("\n" + "="*70)
print("STEG 8: Generera AI-analys")
print("="*70)

llm_analysis = "AI-analys inte tillgänglig (ingen Groq API-nyckel)"
if groq_api_key:
    try:
        from groq import Groq
        groq_client = Groq(api_key=groq_api_key)
        
        analysis_prompt = f"""
Du är en expert på maskininlärning. Analysera denna Random Forest-modell:

MODELLPRESTANDA:
- Precision: {precision:.3f} ({precision*100:.1f}%)
- Recall: {recall:.3f} ({recall*100:.1f}%)
- F1-score: {f1:.3f}
- Confusion Matrix: TP={tp}, FP={fp}, FN={fn}, TN={tn}

HISTORIK:
- Senaste F1: {latest_f1 if latest_f1 else 'Ingen tidigare'}
- Förändring: {change_percent:+.2f}%
- Status: {status}

Skriv en KORT analys (max 3 meningar) på svenska med rekommendation.
"""
        
        response = groq_client.chat.completions.create(
            model="llama-3.3-70b-versatile",
            messages=[{"role": "user", "content": analysis_prompt}],
            temperature=0.5
        )
        llm_analysis = response.choices[0].message.content
        print("   AI-analys genererad!")
    except Exception as e:
        print(f"   Kunde inte generera AI-analys: {e}")

# =============================================
# Steg 9: Slack-rapport (om webhook finns)
# =============================================
print("\n" + "="*70)
print("STEG 9: Slack-rapport")
print("="*70)

if slack_webhook_url:
    slack_message = f"""{status} INSIDER THREAT MODELL – RAPPORT

PRESTANDA:
- Precision: {precision*100:.1f}%
- Recall: {recall*100:.1f}%
- F1-score: {f1:.3f}
- Status: {status}

HISTORIK:
- Senaste F1: {latest_f1 if latest_f1 else 'Ingen tidigare'}
- Forandring: {change_percent:+.2f}%
- Antal körningar: {len(historical_f1_list)}

AI-ANALYS:
{llm_analysis[:300]}

Tid: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}
"""

    try:
        response = requests.post(slack_webhook_url, json={"text": slack_message})
        if response.status_code == 200:
            print("   Slack-rapport skickad!")
        else:
            print(f"   Slack-fel: {response.status_code}")
    except Exception as e:
        print(f"   Kunde inte skicka: {e}")
else:
    print("   Hoppar over Slack (ingen webhook angiven)")

# =============================================
# Steg 10: Logga till MLflow
# =============================================
print("\n" + "="*70)
print("STEG 10: Logga till MLflow")
print("="*70)

run_name = f"Automated_Job_{datetime.now().strftime('%Y%m%d_%H%M%S')}"

with mlflow.start_run(run_name=run_name):
    mlflow.log_param("n_estimators", n_estimators)
    mlflow.log_param("max_depth", max_depth)
    mlflow.log_metric("precision", precision)
    mlflow.log_metric("recall", recall)
    mlflow.log_metric("f1_score", f1)
    mlflow.log_metric("false_positives", int(fp))
    mlflow.log_metric("false_negatives", int(fn))
    mlflow.log_metric("true_positives", int(tp))
    mlflow.log_metric("true_negatives", int(tn))
    
    mlflow.sklearn.log_model(model, "random_forest_model")
    joblib.dump(scaler, "scaler.pkl")
    mlflow.log_artifact("scaler.pkl")
    mlflow.set_tag("job_type", "automated_monitoring")
    mlflow.set_tag("user_email", user_email)

print(f"   Loggat som: {run_name}")

# =============================================
# Steg 11: Spara till CSV
# =============================================
print("\n" + "="*70)
print("STEG 11: Spara historik")
print("="*70)

history_file = "model_performance_history.csv"
performance_record = pd.DataFrame([{
    'timestamp': datetime.now(),
    'precision': precision,
    'recall': recall,
    'f1_score': f1,
    'n_estimators': n_estimators,
    'max_depth': max_depth,
    'change_percent': change_percent,
    'status': status,
    'run_id': mlflow.active_run().info.run_id if mlflow.active_run() else 'N/A'
}])

try:
    if os.path.exists(history_file):
        existing = pd.read_csv(history_file)
        updated = pd.concat([existing, performance_record], ignore_index=True)
        updated.to_csv(history_file, index=False)
        print(f"   Historik uppdaterad ({len(updated)} körningar)")
    else:
        performance_record.to_csv(history_file, index=False)
        print("   Ny historikfil skapad")
except Exception as e:
    print(f"   Kunde inte spara historik: {e}")

# =============================================
# Steg 12: Slutrapport
# =============================================
print("\n" + "="*70)
print("SLUTRAPPORT")
print("="*70)
print(f"""
AUTOMATISERINGSJOB – SAMMANFATTNING

Datum/Tid: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}
Modell: Random Forest (n_estimators={n_estimators}, max_depth={max_depth})

PRESTANDA:
- Precision: {precision*100:.1f}%
- Recall: {recall*100:.1f}%
- F1-score: {f1:.3f}

CONFUSION MATRIX:
- TP: {tp} | FP: {fp}
- FN: {fn} | TN: {tn}

HISTORIK:
- Status: {status}
- Forandring: {change_percent:+.2f}%
- Totalt körningar: {len(historical_f1_list) + 1}

LEVERANSER:
- Slack: {'Skickad' if slack_webhook_url else 'Hoppades over'}
- MLflow: Sparad
- CSV: {'Sparad' if os.path.exists(history_file) else 'Fel'}
""")

print("\n" + "="*70)
print("FULL AUTOMATISERING KLAR!")
print("="*70)
```

```python
# =============================================
# DEL 1: LADDA NER MODELL FRAN MLflow OCH SKAPA ZIP-PAKET
# Korrigerad version - fungerar med array-fel fixat
# =============================================

print("="*70)
print("LADDA NER MODELL FRAN MLflow OCH SKAPA ZIP-PAKET")
print("="*70)

print("""
FORSTA DENNA KOD:

VAD GOR DEN?
- Hamtar en sparad modell från MLflow
- Skapar ett ZIP-paket med modellen, scaler (om den finns) och README
- Låter dig ladda ner ZIP-paketet till din dator

VARFOR BEHOVS DEN?
- MLflow lagrar modeller i Databricks (du kan inte ta med dem)
- ZIP-paketet låter dig spara modellen lokalt på din dator
- Du kan dela modellen med andra (via mail, GitHub, etc.)

HUR FUNGERAR DEN?
1. Listar alla dina sparade modeller i MLflow
2. Du valjer vilken modell du vill ladda ner
3. Koden skapar ett ZIP-paket med modellen, scaler och README
4. Du kan ladda ner ZIP-paketet via en länk
""")

input("Tryck Enter for att fortsatta...")

import mlflow
import joblib
import os
import zipfile
import tempfile
from datetime import datetime
import pandas as pd
import numpy as np
from getpass import getpass
from mlflow.tracking import MlflowClient

client = MlflowClient()

# =============================================
# DEL 1: HITTA OCH VALJ MODELL
# =============================================
print("\n" + "="*70)
print("DEL 1: Hitta och valj modell")
print("="*70)

print("""
FORKLARING:
- Koden letar efter alla modeller du har sparat i MLflow
- Varje modell har ett namn, Run ID, F1-score och status om scaler finns
- Valj en modell som har [OK] Scaler for att den ska fungera direkt
""")

print("\nAnge din e-postadress (samma som i MLflow):")
user_email = input("   -> ").strip()

print("\nAnge experiment-namn (tryck Enter for 'insider_threat_project'):")
exp_name = input("   -> ").strip()
if not exp_name:
    exp_name = "insider_threat_project"

experiment_name = f"/Users/{user_email}/{exp_name}"
print(f"\n   Sokker efter experiment: {experiment_name}")

# Hitta experimentet
experiment = client.get_experiment_by_name(experiment_name)

if not experiment:
    print(f"\n[FEL] Hittade inte experimentet: {experiment_name}")
    raise SystemExit("Experiment saknas")

# Hamta alla korningar
runs = mlflow.search_runs(
    experiment_ids=[experiment.experiment_id],
    order_by=["start_time DESC"]
)

if len(runs) == 0:
    print("\n[FEL] Inga modeller hittades!")
    raise SystemExit("Inga modeller hittades")

print(f"\n[OK] Hittade {len(runs)} sparade modeller:\n")
print("   " + "-"*65)

# Visa alla modeller med information
model_list = []
for i, row in runs.iterrows():
    run_id = row['run_id']
    run_name = row.get('tags.mlflow.runName', 'Unnamed')
    f1_val = row.get('metrics.f1_score', None)
    f1_str = f"{f1_val:.3f}" if f1_val and not pd.isna(f1_val) else "N/A"
    
    # Kontrollera om scaler finns
    has_scaler = False
    n_features = None
    try:
        artifacts = mlflow.artifacts.list_artifacts(run_id)
        has_scaler = any('scaler' in a.path for a in artifacts)
        
        # Kolla antal features
        model_test = mlflow.sklearn.load_model(f"runs:/{run_id}/random_forest_model")
        n_features = model_test.n_features_in_
    except:
        pass
    
    # Bestam status
    if has_scaler and n_features == 14:
        status = "[OK] Scaler finns, 14 features"
    elif has_scaler:
        status = "[VARNING] Scaler finns men fel antal features"
    elif n_features == 14:
        status = "[VARNING] 14 features men ingen scaler"
    else:
        status = "[FEL] Saknar scaler och/eller fel features"
    
    print(f"   {i+1}. {run_name[:40]}")
    print(f"      Run ID: {run_id[:8]}... | F1: {f1_str}")
    print(f"      Status: {status}")
    print()
    
    model_list.append({
        'index': i + 1,
        'run_id': run_id,
        'name': run_name,
        'f1': f1_val,
        'has_scaler': has_scaler,
        'n_features': n_features
    })

# =============================================
# DEL 2: VALJ MODELL ATT LADDA NER
# =============================================
print("\n" + "="*70)
print("DEL 2: Valj modell att ladda ner")
print("="*70)

print("\nValj modell att ladda ner (ange nummer 1-{}):".format(len(model_list)))
choice = input("   -> ").strip()

if choice.isdigit() and 1 <= int(choice) <= len(model_list):
    selected = model_list[int(choice)-1]
    RUN_ID = selected['run_id']
    run_name = selected['name']
    print(f"\n[OK] Valde: {run_name[:50]}")
else:
    RUN_ID = model_list[0]['run_id']
    print(f"\n[OK] Anvander senaste modellen: {model_list[0]['name'][:50]}")

print(f"\n   Run ID: {RUN_ID[:8]}...")

# =============================================
# DEL 3: LADDA MODELL OCH SCALER
# =============================================
print("\n" + "="*70)
print("DEL 3: Ladda modell och scaler från MLflow")
print("="*70)

print("""
FORKLARING OM SCALER:
- MLflow sparar ofta INTE scalern tillsammans med modellen
- Detta är ett kant problem i MLflow
- Om scaler saknas kan du anda anvanda modellen om du skapar en ny scaler sjalv
""")

model = None
scaler = None

try:
    # Ladda modellen
    print("   Laddar modell...", end=" ", flush=True)
    model = mlflow.sklearn.load_model(f"runs:/{RUN_ID}/random_forest_model")
    print("[OK]")
    
    # Ladda scalern (om den finns)
    print("   Laddar scaler...", end=" ", flush=True)
    with tempfile.TemporaryDirectory() as tmp_dir:
        artifacts = mlflow.artifacts.list_artifacts(RUN_ID)
        scaler_found = False
        
        for artifact in artifacts:
            if 'scaler' in artifact.path:
                local_path = mlflow.artifacts.download_artifacts(
                    run_id=RUN_ID,
                    artifact_path=artifact.path,
                    dst_path=tmp_dir
                )
                scaler = joblib.load(os.path.join(local_path, artifact.path))
                print(f"[OK] (fanns i MLflow)")
                scaler_found = True
                break
        
        if not scaler_found:
            print("[VARNING] (ingen scaler i MLflow)")
            
            # Forsok anvanda scaler från workspace
            if 'scaler' in globals() and globals()['scaler'] is not None:
                scaler = globals()['scaler']
                print("   [OK] Anvander scaler från workspace som backup")
            else:
                print("   [INFO] Ingen scaler tillganglig - du maste skapa en ny sjalv")
    
except Exception as e:
    print(f"\n[FEL] Kunde inte ladda modell: {e}")
    raise

# =============================================
# DEL 4: HAMTA MODELLENS PRESTANDA
# =============================================
print("\n" + "="*70)
print("DEL 4: Modellens prestanda")
print("="*70)

try:
    run = mlflow.get_run(RUN_ID)
    metrics = run.data.metrics
    params = run.data.params
    
    precision = metrics.get('precision', 0.991)
    recall = metrics.get('recall', 0.967)
    f1 = metrics.get('f1_score', 0.979)
    n_estimators = params.get('n_estimators', '100')
    max_depth = params.get('max_depth', '10')
    
    print(f"\n   Precision: {float(precision)*100:.1f}%")
    print(f"   Recall: {float(recall)*100:.1f}%")
    print(f"   F1-score: {float(f1):.3f}")
    print(f"   n_estimators: {n_estimators}, max_depth: {max_depth}")
except:
    precision, recall, f1 = 0.991, 0.967, 0.979
    n_estimators, max_depth = 100, 10
    print("\n   [VARNING] Kunde inte hamta metrics, anvander standardvarden")

# =============================================
# DEL 5: VALJ SPARPLATS FOR ZIP-PAKET
# =============================================
print("\n" + "="*70)
print("DEL 5: Valj var ZIP-paketet ska sparas")
print("="*70)

print("""
FORKLARING:
- /tmp: Tillfallig plats (forsvinner nar du stanger Databricks)
- /Workspace: Permanent plats (finns kvar i din workspace)
- Bada: Sparar pa bada stallen for extra sakerhet
""")

print("\n   Var vill du spara ZIP-paketet?")
print("      1. /tmp (tillfallig)")
print("      2. /Workspace (permanent)")
print("      3. Bada stallen")

location_choice = input("   -> ").strip()

# =============================================
# DEL 6: SKAPA TEMPORAR MAPP
# =============================================
print("\n" + "="*70)
print("DEL 6: Skapa ZIP-paket")
print("="*70)

timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
temp_dir = f"/tmp/model_export_{timestamp}"
os.makedirs(temp_dir, exist_ok=True)

# Spara modell
model_temp_path = os.path.join(temp_dir, "random_forest_model.pkl")
joblib.dump(model, model_temp_path)
print(f"   Modell sparad: {model_temp_path}")

# Spara scaler (om den finns)
if scaler:
    scaler_temp_path = os.path.join(temp_dir, "scaler.pkl")
    joblib.dump(scaler, scaler_temp_path)
    print(f"   Scaler sparad: {scaler_temp_path}")
else:
    print("   Ingen scaler att spara")

# =============================================
# DEL 7: SKAPA README-FIL (KORRIGERAD)
# =============================================
print("\n   Skapar README-fil med instruktioner...")

# Skapa en enkel sträng för prediktion (utan array-problem)
prediction_example = """# Gor forutsagelse
X_scaled = scaler.transform(X_new)
pred = model.predict(X_scaled)[0]
if pred == 1:
    print("ATTACK upptackt!")
else:
    print("NORMAL aktivitet")"""

readme_content = f"""MODEL BACKUP – Insider Threat Detection
========================================

Exportdatum: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}
Run ID: {RUN_ID}
Anvandare: {user_email}

MODELLINFO:
- Typ: Random Forest Classifier
- n_estimators: {n_estimators}
- max_depth: {max_depth}
- Precision: {float(precision)*100:.1f}%
- Recall: {float(recall)*100:.1f}%
- F1-score: {float(f1):.3f}

FILER I DETTA PAKET:
- random_forest_model.pkl : Tranad Random Forest-modell
- scaler.pkl : StandardScaler for att skala indata (om tillganglig)

FOR ATT LADDA I PYTHON:
------------------------------------------------
import joblib
import pandas as pd

# Ladda modell och scaler
model = joblib.load('random_forest_model.pkl')
if scaler_exists:
    scaler = joblib.load('scaler.pkl')

# Forbered indata (maste ha samma 14 features)
X_new = pd.DataFrame([{{
    'NUM_FAILED_LOGINS': 0,
    'ROOT_SHELL': 0,
    'SU_ATTEMPTED': 0,
    'NUM_ACCESS_FILES': 5,
    'NUM_FILE_CREATIONS': 1,
    'SRC_BYTES': 100,
    'DST_BYTES': 200,
    'DURATION': 300,
    'SAME_SRV_RATE': 0.8,
    'DIFF_SRV_RATE': 0.2,
    'LOGGED_IN': 1,
    'TOTAL_PRIVILEGE_ATTEMPTS': 0,
    'TOTAL_BYTES_TRANSFERRED': 300,
    'FLAG_RISK_SCORE': 0
}}])

# Gör forutsagelse (om du har scaler)
# X_scaled = scaler.transform(X_new)
# prediction = model.predict(X_scaled)[0]
# print("ATTACK" if prediction == 1 else "NORMAL")
------------------------------------------------

Detta paket ar skapat for backup och distribution.
"""

readme_path = os.path.join(temp_dir, "README_BACKUP.txt")
with open(readme_path, 'w', encoding='utf-8') as f:
    f.write(readme_content)
print("   README_BACKUP.txt skapad")

# =============================================
# DEL 8: SKAPA ZIP-FIL
# =============================================
print("\n   Skapar ZIP-paket...")

zip_filename = f"model_backup_{timestamp}.zip"
zip_path_tmp = f"/tmp/{zip_filename}"

with zipfile.ZipFile(zip_path_tmp, 'w', zipfile.ZIP_DEFLATED) as zipf:
    zipf.write(model_temp_path, arcname="random_forest_model.pkl")
    if scaler:
        zipf.write(scaler_temp_path, arcname="scaler.pkl")
    zipf.write(readme_path, arcname="README_BACKUP.txt")

print(f"   ZIP-paket skapat: {zip_path_tmp}")
print(f"   Filstorlek: {os.path.getsize(zip_path_tmp) / 1024:.1f} KB")

# =============================================
# DEL 9: SPARA TILL VALD PLATS
# =============================================
print("\n" + "="*70)
print("DEL 9: Spara ZIP-paket")
print("="*70)

zip_paths = []

if location_choice in ['1', '3']:
    zip_paths.append(("TEMP (/tmp)", zip_path_tmp))
    print(f"[OK] Sparat i /tmp: {zip_filename}")

if location_choice in ['2', '3']:
    try:
        workspace_dir = f"/Workspace/Users/{user_email}/model_backups"
        os.makedirs(workspace_dir, exist_ok=True)
        workspace_zip = os.path.join(workspace_dir, zip_filename)
        
        import shutil
        shutil.copy2(zip_path_tmp, workspace_zip)
        zip_paths.append(("WORKSPACE", workspace_zip))
        print(f"[OK] Sparat i Workspace: {workspace_dir}")
    except Exception as e:
        print(f"[VARNING] Kunde inte spara till Workspace: {e}")

# =============================================
# DEL 10: SKAPA NEDLADDNINGSLANK
# =============================================
print("\n" + "="*70)
print("DEL 10: Ladda ner ZIP-paketet")
print("="*70)

try:
    from IPython.display import display, HTML
    
    display(HTML(f"""
    <div style="background-color: #4CAF50; padding: 15px; border-radius: 10px; text-align: center; margin: 10px;">
        <a href="files/tmp/{zip_filename}" download="{zip_filename}" 
           style="color: white; font-size: 18px; text-decoration: none; font-weight: bold;">
           KLICKA HAR FOR ATT LADDA NER ZIP-PAKETET
        </a>
    </div>
    """))
    print("\n[OK] Klicka pa den grona länken ovan for att ladda ner!")
except Exception as e:
    print(f"\n[INFO] Manuell nedladdning kravs:")
    print(f"   1. Ga till 'Data' -> 'DBFS' i vanstermenyn")
    print(f"   2. Navigera till: /tmp/")
    print(f"   3. Hitta filen: {zip_filename}")
    print("   4. Hogerklicka -> 'Download'")

# =============================================
# SLUTSAMMANFATTNING
# =============================================
print("\n" + "="*70)
print("SAMMANFATTNING")
print("="*70)

print(f"""
[OK] ZIP-PAKETET AR KLART!

Run ID: {RUN_ID[:8]}...
Datum: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}

ZIP-PAKETET INNEHALLER:
- random_forest_model.pkl (den tranade modellen)
- {'scaler.pkl (StandardScaler)' if scaler else 'scaler.pkl (saknas)'}
- README_BACKUP.txt (instruktioner)

FILNAMN: {zip_filename}
STORLEK: {os.path.getsize(zip_path_tmp) / 1024:.1f} KB

SPARAD PA:
""" + "\n".join([f"   - {loc}: {path}" for loc, path in zip_paths]) + f"""

NASTA STEG:
1. Ladda ner ZIP-paketet (klicka pa den grona länken)
2. Packa upp ZIP-filen
3. Anvand modellen for prediktioner!

[KLAR] ZIP-paketet ar redo for nedladdning!
""")

print("\n" + "="*70)
print("KLART!")
print("="*70)
```
